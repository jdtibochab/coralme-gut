```python
# !bash report.sh
organisms = set(pandas.read_csv("biomass_constrained.txt",index_col=0,header=None).index.to_list())
done = {i.split(".csv")[0] for i in os.listdir("./datasets/m_subsystems/")}
```

```python
run_for = organisms - done
len(run_for)
```

```python
import cobra

def run(org):
    model = cobra.io.json.load_json_model('./agora-models/AGORA_2_01_json_renamed/{}.json'.format(org))
    subsystems = pandas.Series({r.id:r.subsystem for r in model.reactions})
    subsystems.to_csv("datasets/m_subsystems/{}.csv".format(org))
```

```python
NP = min([10,len(run_for)])
pool = mp.Pool(NP,maxtasksperchild=1)
pbar = tqdm(total=len(run_for),position=0,leave=True)
pbar.set_description('Getting ({} threads)'.format(NP))
def collect_result(result):
    pbar.update(1)
    
for org in run_for:
    args = ([org])
    pool.apply_async(run,args, callback=collect_result)
pool.close()
pool.join()
```
